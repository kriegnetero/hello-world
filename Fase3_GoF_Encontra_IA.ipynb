## 📖 Passo 1 — Os 18 Padrões do Catálogo CSIRO

O artigo organiza os padrões em **6 grupos funcionais**:

| Grupo | Padrões |
|-------|---------|
| **Goal Acquisition** | Passive Goal Creator, Proactive Goal Creator |
| **Prompt Optimization** | Prompt/Response Optimiser |
| **Knowledge Augmentation** | RAG, One-Shot Model Querying, Incremental Model Querying |
| **Plan Generation** | Single-Path Plan Generator, Multi-Path Plan Generator |
| **Reflection** | Self-Reflection, Cross-Reflection, Human Reflection |
| **Multi-Agent Cooperation** | Voting-based, Role-based, Debate-based Cooperation |
| **Infrastructure** | Multimodal Guardrails, Tool/Agent Registry, Agent Adapter, Agent Evaluator |

Cada padrão é documentado com: *Context, Problem, Forces, Solution, Consequences, Known Uses, Related Patterns* — exatamente o formato GoF!
