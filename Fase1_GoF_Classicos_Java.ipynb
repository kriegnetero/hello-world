## 🔹 Passo 3 — Singleton (Criacional)

### Problema
Garantir que uma classe tenha **exatamente uma instância** e fornecer um ponto global de acesso.

### Quando usar
- Configurações globais da aplicação
- Pool de conexões de banco de dados
- Logger centralizado

### Estrutura UML
```
┌──────────────────┐
│    Singleton      │
├──────────────────┤
│ - instance: Singleton │
├──────────────────┤
│ - Singleton()         │
│ + getInstance(): Singleton │
│ + operacao(): void    │
└──────────────────┘
```


````java
// ══ SINGLETON — Configuração do Agente de IA ══
// Contexto: em um sistema de agentes IA, a configuração
// (API key, modelo, temperatura) deve ser única e compartilhada.

public class AgentConfig {
    // Instância única (Lazy Initialization + Thread-Safe)
    private static volatile AgentConfig instance;
    
    // Atributos de configuração
    private String modelName;
    private double temperature;
    private String apiKey;
    
    // Construtor PRIVADO — impede new AgentConfig()
    private AgentConfig() {
        this.modelName = "gpt-4o-mini";
        this.temperature = 0.7;
        this.apiKey = "sk-***";
        System.out.println("⚙️ AgentConfig criado (única vez!)");
    }
    
    // Ponto de acesso global (Double-Checked Locking)
    public static AgentConfig getInstance() {
        if (instance == null) {
            synchronized (AgentConfig.class) {
                if (instance == null) {
                    instance = new AgentConfig();
                }
            }
        }
        return instance;
    }
    
    // Getters e setters
    public String getModelName() { return modelName; }
    public void setModelName(String m) { this.modelName = m; }
    public double getTemperature() { return temperature; }
    
    @Override
    public String toString() {
        return "AgentConfig{model=" + modelName + ", temp=" + temperature + "}";
    }
}

// ── Teste ──
AgentConfig c1 = AgentConfig.getInstance();
AgentConfig c2 = AgentConfig.getInstance();
System.out.println("c1: " + c1);
System.out.println("c2: " + c2);
System.out.println("c1 == c2? " + (c1 == c2));  // true — mesma instância!
c1.setModelName("claude-sonnet-4");
System.out.println("c2 após mudar c1: " + c2.getModelName());  // claude-sonnet-4!
````

## 🔹 Passo 4 — Factory Method (Criacional)

### Problema
Definir uma interface para criar objetos, mas deixar que **subclasses decidam** qual classe instanciar.

### Quando usar
- Sistemas de plugins
- Frameworks que precisam instanciar objetos sem conhecer a classe concreta
- Criação de agentes de diferentes tipos

### Conexão com o artigo CSIRO
O **Passive Goal Creator** e o **Proactive Goal Creator** são dois tipos de criadores de objetivo — um Factory Method pode decidir qual instanciar com base no contexto.


````java
// ══ FACTORY METHOD — Criação de Agentes por Tipo ══

// Produto abstrato
interface Agent {
    String processGoal(String userInput);
    String getType();
}

// Produtos concretos
class PassiveGoalAgent implements Agent {
    public String processGoal(String input) {
        return "📝 [Passive] Analisando prompt direto: '" + input + "'";
    }
    public String getType() { return "Passive Goal Creator"; }
}

class ProactiveGoalAgent implements Agent {
    public String processGoal(String input) {
        return "🔍 [Proactive] Capturando contexto e antecipando objetivo: '" + input + "'";
    }
    public String getType() { return "Proactive Goal Creator"; }
}

// Factory Method — o criador abstrato
abstract class AgentFactory {
    // Factory Method — subclasses implementam
    abstract Agent createAgent();
    
    // Template que usa o factory method
    public String executeGoal(String input) {
        Agent agent = createAgent();
        System.out.println("🏭 Factory criou: " + agent.getType());
        return agent.processGoal(input);
    }
}

// Fábricas concretas
class PassiveAgentFactory extends AgentFactory {
    Agent createAgent() { return new PassiveGoalAgent(); }
}

class ProactiveAgentFactory extends AgentFactory {
    Agent createAgent() { return new ProactiveGoalAgent(); }
}

// ── Teste ──
AgentFactory factory1 = new PassiveAgentFactory();
System.out.println(factory1.executeGoal("Resuma este documento"));

System.out.println();

AgentFactory factory2 = new ProactiveAgentFactory();
System.out.println(factory2.executeGoal("Abrir câmera e analisar ambiente"));
````

## 🔹 Passo 5 — Observer (Comportamental)

### Problema
Definir uma dependência **um-para-muitos** entre objetos, de modo que quando um muda de estado, todos os dependentes são notificados.

### Conexão com o artigo CSIRO
O padrão **Voting-based Cooperation** usa Observer: quando um agente emite seu voto, os demais agentes e o coordenador são notificados para contabilizar e reagir.


````java
// ══ OBSERVER — Sistema de Votação de Agentes ══
import java.util.*;

// Subject (Observable)
interface VotingSubject {
    void registerObserver(VoteObserver o);
    void removeObserver(VoteObserver o);
    void notifyObservers(String agentName, String vote);
}

// Observer
interface VoteObserver {
    void onVoteReceived(String agentName, String vote);
}

// ConcreteSubject — Coordenador de votação
class VotingCoordinator implements VotingSubject {
    private List<VoteObserver> observers = new ArrayList<>();
    private Map<String, String> votes = new HashMap<>();
    
    public void registerObserver(VoteObserver o) { observers.add(o); }
    public void removeObserver(VoteObserver o) { observers.remove(o); }
    
    public void notifyObservers(String agentName, String vote) {
        for (VoteObserver o : observers) {
            o.onVoteReceived(agentName, vote);
        }
    }
    
    public void submitVote(String agentName, String vote) {
        votes.put(agentName, vote);
        System.out.println("🗳️ " + agentName + " votou: " + vote);
        notifyObservers(agentName, vote);
    }
    
    public Map<String, String> getResults() { return votes; }
}

// ConcreteObservers
class AuditLogger implements VoteObserver {
    public void onVoteReceived(String agent, String vote) {
        System.out.println("  📋 [AuditLog] Registrado: " + agent + " → " + vote);
    }
}

class ConsensusChecker implements VoteObserver {
    private Map<String, Integer> tally = new HashMap<>();
    public void onVoteReceived(String agent, String vote) {
        tally.merge(vote, 1, Integer::sum);
        System.out.println("  📊 [Consensus] Contagem atual: " + tally);
    }
}

// ── Teste ──
VotingCoordinator coordinator = new VotingCoordinator();
coordinator.registerObserver(new AuditLogger());
coordinator.registerObserver(new ConsensusChecker());

coordinator.submitVote("Agent-A", "Plano Alpha");
coordinator.submitVote("Agent-B", "Plano Beta");
coordinator.submitVote("Agent-C", "Plano Alpha");
System.out.println("\nResultado final: " + coordinator.getResults());
````

## 🔹 Passo 6 — Strategy (Comportamental)

### Problema
Definir uma família de algoritmos, encapsular cada um e torná-los **intercambiáveis**. O cliente escolhe o algoritmo em tempo de execução.

### Conexão com o artigo CSIRO
O **RAG (Retrieval Augmented Generation)** vs **One-Shot Model Querying** vs **Incremental Model Querying** são três *estratégias* diferentes de consulta ao modelo. O padrão Strategy permite trocar de estratégia sem alterar o agente.


````java
// ══ STRATEGY — Estratégias de Consulta ao Modelo ══

// Strategy interface
interface QueryStrategy {
    String query(String prompt, String context);
    String getName();
}

// ConcreteStrategy A — One-Shot (uma chamada ao modelo)
class OneShotQueryStrategy implements QueryStrategy {
    public String query(String prompt, String context) {
        return "⚡ [One-Shot] Gerou plano completo em 1 chamada para: '" + prompt + "'";
    }
    public String getName() { return "One-Shot Model Querying"; }
}

// ConcreteStrategy B — Incremental (múltiplas chamadas)
class IncrementalQueryStrategy implements QueryStrategy {
    public String query(String prompt, String context) {
        StringBuilder sb = new StringBuilder();
        sb.append("🔄 [Incremental] Passo 1: Analisar '" + prompt + "'\n");
        sb.append("   Passo 2: Gerar sub-tarefas\n");
        sb.append("   Passo 3: Refinar com contexto: '" + context + "'");
        return sb.toString();
    }
    public String getName() { return "Incremental Model Querying"; }
}

// ConcreteStrategy C — RAG (busca + geração)
class RAGQueryStrategy implements QueryStrategy {
    public String query(String prompt, String context) {
        return "📚 [RAG] Buscou docs relevantes → Aumentou prompt → Gerou resposta para: '" + prompt + "'";
    }
    public String getName() { return "Retrieval Augmented Generation"; }
}

// Context — o Agente que usa a Strategy
class AgentQueryEngine {
    private QueryStrategy strategy;
    
    public void setStrategy(QueryStrategy s) {
        this.strategy = s;
        System.out.println("🔧 Estratégia trocada para: " + s.getName());
    }
    
    public String executeQuery(String prompt, String context) {
        if (strategy == null) throw new IllegalStateException("Nenhuma strategy definida!");
        return strategy.query(prompt, context);
    }
}

// ── Teste — troca de estratégia em runtime ──
AgentQueryEngine engine = new AgentQueryEngine();

engine.setStrategy(new OneShotQueryStrategy());
System.out.println(engine.executeQuery("Resuma o artigo", ""));

System.out.println();

engine.setStrategy(new RAGQueryStrategy());
System.out.println(engine.executeQuery("Quais são os 18 padrões?", "artigo_csiro.pdf"));

System.out.println();

engine.setStrategy(new IncrementalQueryStrategy());
System.out.println(engine.executeQuery("Crie um plano de testes", "specs.md"));
````

## 🔹 Passo 7 — Adapter (Estrutural)

### Problema
Converter a interface de uma classe em outra interface que o cliente espera. Permite que classes com interfaces incompatíveis trabalhem juntas.

### Conexão com o artigo CSIRO
O **Agent Adapter** (padrão 17 do catálogo) é literalmente o Adapter GoF aplicado a agentes de IA: conecta o agente a ferramentas externas com APIs diferentes.


````java
// ══ ADAPTER — Conectando Agente a Ferramentas Externas ══

// Interface esperada pelo Agente (Target)
interface ToolInterface {
    String execute(String input);
    String getToolName();
}

// Ferramenta legada 1 — API incompatível (Adaptee)
class LegacyDatabaseSearch {
    public String searchRecords(String query, int maxResults, boolean fuzzy) {
        return "🗄️ DB encontrou " + maxResults + " registros para '" + query + "'";
    }
}

// Ferramenta legada 2 — API diferente (Adaptee)
class ExternalWebScraper {
    public String fetchPage(String url) {
        return "🌐 Conteúdo extraído de: " + url;
    }
    public String extractText(String html) {
        return "📄 Texto limpo extraído";
    }
}

// Adapter 1 — adapta LegacyDatabaseSearch para ToolInterface
class DatabaseSearchAdapter implements ToolInterface {
    private LegacyDatabaseSearch legacyDB;
    
    public DatabaseSearchAdapter(LegacyDatabaseSearch db) {
        this.legacyDB = db;
    }
    
    public String execute(String input) {
        // Adapta a interface: 1 parâmetro → 3 parâmetros
        return legacyDB.searchRecords(input, 10, true);
    }
    public String getToolName() { return "Database Search (Adapted)"; }
}

// Adapter 2 — adapta ExternalWebScraper para ToolInterface
class WebScraperAdapter implements ToolInterface {
    private ExternalWebScraper scraper;
    
    public WebScraperAdapter(ExternalWebScraper ws) {
        this.scraper = ws;
    }
    
    public String execute(String input) {
        // Orquestra duas chamadas da API legada
        String html = scraper.fetchPage(input);
        return scraper.extractText(html);
    }
    public String getToolName() { return "Web Scraper (Adapted)"; }
}

// ── Teste — o Agente usa ferramentas via interface unificada ──
List<ToolInterface> tools = List.of(
    new DatabaseSearchAdapter(new LegacyDatabaseSearch()),
    new WebScraperAdapter(new ExternalWebScraper())
);

String userQuery = "design patterns AI agents";
System.out.println("🤖 Agente processando: '" + userQuery + "'\n");
for (ToolInterface tool : tools) {
    System.out.println("  Usando " + tool.getToolName() + ":");
    System.out.println("  → " + tool.execute(userQuery));
    System.out.println();
}
System.out.println("💡 O Agente não conhece as APIs legadas — só usa ToolInterface!");
````
